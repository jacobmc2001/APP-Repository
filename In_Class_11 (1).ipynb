# Exercise 2 (15 points)

In this exercise, we will practice documenting code the way Chapter 9 describes: with a descriptive name, type hints, and a complete Google-style docstring (summary, `Args`, `Returns`, `Raises`, and `Example`).

Consider the following undocumented, poorly named function:

```python
def f(df, n):
    df = df[df['price'] >= n]
    df['price'] = df['price'].astype(float)
    return df
```

Your task:

- **Rename the function** to something descriptive that follows the `verb_noun` convention (e.g., a function that filters rows should start with a verb like `filter_`).
- **Add type hints** to the parameters and the return value (the input and output are `pandas.DataFrame`; the threshold is a `float`).
- **Add a `ValueError`** that is raised if the threshold `n` is negative.
- **Write a complete Google-style docstring** including a one-line summary, `Args`, `Returns`, `Raises`, and a runnable `Example`.

Then apply your documented function to the following data, using a minimum price threshold of `20`.

```python
import pandas as pd

data = {'item': ['A', 'B', 'C', 'D'],
        'price': ['10', '25', '30', '15']}
df = pd.DataFrame(data)
```

The expected outcome is shown below.

| item | price |
| --- | --- |
| B | 25.0 |
| C | 30.0 |

````python

import pandas as pd

data = {
    'item': ['A', 'B', 'C', 'D'],
    'price': ['10', '25', '30', '15']
}

df = pd.DataFrame(data)

print(df)
````

````python

def filter_by_price(df: pd.DataFrame, threshold: float) -> pd.DataFrame:
    """Filter items by a minimum price.

    Args:
        df: A DataFrame containing item prices.
        threshold: The minimum price to keep.

    Returns:
        A DataFrame with prices at or above the threshold.

    Raises:
        ValueError: If the threshold is negative.

    Example:
        >>> data = pd.DataFrame({'price': ['10', '25', '30']})
        >>> filter_by_price(data, 20)['price'].tolist()
        [25.0, 30.0]
    """
    if threshold < 0:
        raise ValueError("Threshold cannot be negative.")

    result = df.copy()
    result['price'] = result['price'].astype(float)
    return result[result['price'] >= threshold]
````

````python

filtered_df = filter_by_price(df, 20)

print(filtered_df)
````

````python

try:
    filter_by_price(df, -10)
except ValueError as e:
    print("Error caught:", e)
````

````python

assert filtered_df['item'].tolist() == ['B', 'C']
assert filtered_df['price'].tolist() == [25.0, 30.0]

print("All tests passed!")
````
