# Exercise 3 (8 points)

How many products cost more than $10.00?

```python
chipotle["item_price"] = chipotle["item_price"].str.replace("$", "").astype(float)

chipotle[chipotle["item_price"] > 10]["item_name"].nunique()
```

# Exercise 4 (2 points)

Sort by the name of the item

```python
chipotle.sort_values(by="item_name")
```

# Exercise 5 (2 points)

What was the quantity of the most expensive item ordered?

```python
chipotle.loc[chipotle["item_price"].idxmax(), "quantity"]
```

# Exercise 6 (2 points)

How many times was a Veggie Salad Bowl ordered?

```python
chipotle[chipotle["item_name"] == "Veggie Salad Bowl"]["quantity"].sum()
```

# Exercise 7 (3 points)

How many times did someone order more than one Canned Soda?

```python
chipotle[(chipotle["item_name"] == "Canned Soda") & (chipotle["quantity"] > 1)].shape[0]
```
