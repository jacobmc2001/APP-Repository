### Exercise 1(e) (6 points)

Convert the `cut`, `color`, and `clarity` columns to 0-1 columns, and append them to the `diamonds` data frame. Then, drop `cut`, `color`, and `clarity` from the data frame.

```python
dummy_columns = pd.get_dummies(diamonds[["cut", "color", "clarity"]], dtype=int)

diamonds = pd.concat([diamonds, dummy_columns], axis=1)

diamonds = diamonds.drop(columns=["cut", "color", "clarity"])
```

### Exercise 1(f) (4 points)

Create a lambda function called  `ratio` that computes the ratio of `depth` and `table`. Then, create a column called `depth/table`, using the `apply` function, that stores the values of `ratio` applied to the `diamonds` data frame.

```python
ratio = lambda depth, table: depth / table

diamonds["depth/table"] = diamonds.apply(
    lambda row: ratio(row["depth"], row["table"]),
    axis=1
)
```

### Exercise 1(g) (4 points)

Create a lambda function called `vol` that computes the product of `x`, `y`, and `z`. Then, create a column called `volume`, using the `apply` function, that stores the values of `vol` applued to the `diamonds` data frame.

```python
vol = lambda x, y, z: x * y * z

diamonds["volume"] = diamonds.apply(
    lambda row: vol(row["x"], row["y"], row["z"]),
    axis=1
)
```

### Exercise 2(a) (12 points)

Create a class called `DataProcessing`. The class should include the following methods:

- `__init__(self, data: pd.DataFrame)`: Initializes the class with a pandas DataFrame.
- `encode_categorical(self, columns)`: Encodes categorical columns using `get_dummies` encoding from `pandas`. 
- `volume(self)`: Computes the volume of the diamond as `x * y * z`.
- `ratio(self)`: Computes the ratio of `depth` and `table`.
- `stand(self)`: Computes the z-score standardization of `carat`. That is, `(carat - mean(carat)) / std(carat)`, where `mean` and `std` are the mean and standard deviation, respectively.
- `fe(self)`: Append the `volume` and `ratio` engineered features to the data frame. Also apply `stand`.

Implement the `DataProcessing` class with the above methods and demonstrate its usage with a sample DataFrame.


```python
class DataProcessing:

    def __init__(self, data: pd.DataFrame):
        self.data = data

    def encode_categorical(self, columns):
        encoded = pd.get_dummies(self.data[columns], dtype=int)
        self.data = pd.concat([self.data, encoded], axis=1)
        self.data = self.data.drop(columns=columns)

    def volume(self):
        return self.data["x"] * self.data["y"] * self.data["z"]

    def ratio(self):
        return self.data["depth"] / self.data["table"]

    def stand(self):
        return (self.data["carat"] - self.data["carat"].mean()) / self.data["carat"].std()

    def fe(self):
        self.data["volume"] = self.volume()
        self.data["ratio"] = self.ratio()
        self.data["carat_stand"] = self.stand()
        return self.data
```

For each principle, provide one concrete example using the `DataProcessing` class you built in Exercise 2 (or another pandas/scikit-learn object of your choice).

Encapsulation: Keeping data and the functions that work with it together. In DataProcessing, the data and methods like volume() are all kept in one class.

Abstraction: Making complicated tasks easier to use. For example, fe() runs several calculations for us instead of us doing each one separately.

Inheritance: Letting a new class build off an existing class. For example, a new class could use everything in DataProcessing and add more features.

Polymorphism: Different classes can use the same method name but do different things with it. For example, another class could have its own version of stand().

In this exercise, we will practice inheritance and polymorphism by extending the `DataProcessing` class built in Exercise 2.

### Exercise 4(a) (4 points)

Create a class called `PriceFlagging` that inherits from `DataProcessing`. The class should include a method `flag_price(self, threshold)` that creates a new column called `price_flag`, using `apply` and a lambda function, that stores 1 if `price` is greater than `threshold` and 0 otherwise.

```python
class PriceFlagging(DataProcessing):

    def flag_price(self, threshold):
        self.data["price_flag"] = self.data["price"].apply(
            lambda x: 1 if x > threshold else 0
        )
```

### Exercise 4(b) (4 points)

Read the `diamonds.csv` data file. Initialize the `PriceFlagging` class, run the inherited `fe` method, and then call `flag_price` with a `threshold` of 10000. Display the first five rows of the resulting data frame.

```python
diamonds = pd.read_csv("diamonds.csv")

price_processor = PriceFlagging(diamonds)

price_processor.fe()
price_processor.flag_price(10000)

price_processor.data.head()
```
