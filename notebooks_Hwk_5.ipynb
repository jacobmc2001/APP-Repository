### Exercise 4(a) (3 points)

Import the `logging` and `numpy` libraries. Then, configure logging so that (1) messages are written to a file called `student_analysis.log`, (2) messages with level `INFO` and above are recorded, and (3) each record includes the time, the level, and the message. Use `force=True` so the configuration is not ignored in a notebook. Finally, read the data and create a data frame called `student` (if you have not already done so).

````python

import logging
import numpy as np
import pandas as pd

logging.basicConfig(
    filename="student_analysis.log",
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s",
    force=True
)

student = pd.read_csv("StudentPerformanceFactors.csv")

logging.info("Student data loaded successfully")

print("Dataset loaded successfully!")
print("Rows:", len(student))
````

### Exercise 4(b) (7 points)

Write a function called `column_summary(df, column)` that returns a dictionary with the mean, standard deviation, and median (rounded to 2 decimals) of `column`. The function must:

- Log an `INFO` message when it starts.
- Raise a `ValueError` with an informative message (what was expected, what was received) if `column` is not in `df`.
- Use `try`, `except`, `else`, and `finally` blocks to convert the column to numeric with `pd.to_numeric`. If the column cannot be converted, log the error using `logging.exception` and return a dictionary with the same keys and `np.nan` values (so the return format is consistent). If it can, log an `INFO` message with the summary and return it.
- Log an `INFO` message saying that the processing finished, whether or not an error occurred.

Keep the `try` block as small as possible.

````python

def column_summary(df, column):
    logging.info(f"Starting summary for {column}")

    if column not in df.columns:
        logging.info(f"Finished processing {column} with an error")
        raise ValueError(
            f"Expected column '{column}', but it was not found. "
            f"Available columns: {list(df.columns)}"
        )

    try:
        numeric_data = pd.to_numeric(df[column], errors="raise")

    except (ValueError, TypeError):
        logging.exception(f"Could not process column: {column}")
        return {
            "mean": np.nan,
            "std": np.nan,
            "median": np.nan
        }

    else:
        summary = {
            "mean": round(numeric_data.mean(), 2),
            "std": round(numeric_data.std(), 2),
            "median": round(numeric_data.median(), 2)
        }

        logging.info(f"Summary for {column}: {summary}")
        return summary

    finally:
        logging.info(f"Finished processing {column}")
````

Print the content of `student_analysis.log`. Which log lines were produced by each of the three calls in 4(c)? Which level was used for each?

````python
import logging

logger = logging.getLogger()

# Remove old logging handlers
for handler in logger.handlers[:]:
    logger.removeHandler(handler)
    handler.close()

# Set up a fresh log file
logging.basicConfig(
    filename="student_analysis.log",
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s",
    filemode="w",
    force=True
)

logging.info("Logging has been reset successfully")
````
